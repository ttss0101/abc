```python
# Single Layer Perceptron - AND Gate

w1 = 1
w2 = 1
bias = 0
threshold = 2

def activation(net):
    if net >= threshold:
        return 1
    else:
        return 0

def perceptron(a, b):
    net = (a * w1) + (b * w2) + bias
    return activation(net)

print("AND GATE")
print("A\tB\tOutput")

for a in [0, 1]:
    for b in [0, 1]:
        print(a, "\t", b, "\t", perceptron(a, b))
```

```python
# Single Layer Perceptron - OR Gate

w1 = 1
w2 = 1
bias = 0
threshold = 1

def activation(net):
    if net >= threshold:
        return 1
    else:
        return 0

def perceptron(a, b):
    net = (a * w1) + (b * w2) + bias
    return activation(net)

print("OR GATE")
print("A\tB\tOutput")

for a in [0, 1]:
    for b in [0, 1]:
        print(a, "\t", b, "\t", perceptron(a, b))
```

```python
# ADALINE - AND Gate

import numpy as np

# Input data
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

# Target output
Y = np.array([0, 0, 0, 1])

# Initialize weights and bias
weights = np.zeros(2)
bias = 0

learning_rate = 0.1
epochs = 100

# Training
for epoch in range(epochs):
    for i in range(len(X)):

        net = np.dot(X[i], weights) + bias

        error = Y[i] - net

        weights = weights + learning_rate * error * X[i]

        bias = bias + learning_rate * error

# Testing
print("ADALINE - AND GATE")
print("A\tB\tOutput")

for x in X:

    net = np.dot(x, weights) + bias

    if net >= 0.5:
        output = 1
    else:
        output = 0

    print(x[0], "\t", x[1], "\t", output)

print("Final Weights:", weights)
print("Final Bias:", bias)
```
